```python
from numpy import asarray
from numpy.random import rand
from numpy import arange
import matplotlib.pyplot as plt
import numpy as np
```

```python
def objective(x):
    return x**2.0
```

```python
def derivative(x):
    return x*2.0
```

```python
# gradient descent algorithm
def gradient_descent(objective, derivative, bounds, n_iter, step_size):
    value = []
    eval_value = []
    # generate an initial point
    solution = bounds[:, 0] + rand(len(bounds)) * (bounds[:, 1] - bounds[:, 0])
    # run the gradient descent
    for i in range(n_iter):
        # calculate gradient
        gradient = derivative(solution)
        value.append(solution)
        # take a step
        solution = solution - step_size * gradient
        # evaluate candidate point
        solution_eval = objective(solution)
        eval_value.append(solution_eval)
        # report progress
        print('%d f(%.5f) = %.5f' % (i, solution.item(), solution_eval.item()))#
    return [value, eval_value, solution, solution_eval]
```

```python
# define range for input
bounds = asarray([[-1.0, 1.0]])
# define the total iterations
n_iter = 30
# define the step size
step_size = 0.1
# perform the gradient descent search
value, eval_value, best, score = gradient_descent(objective, derivative, bounds, n_iter, step_size)
print('Done!')
print('f(%.5f) = %.5f' % (best.item(), score.item()))
```

```python
plt.plot(value, eval_value)
```

```python
# define range for input
r_min, r_max = -1.0, 1.0
# sample input range uniformly at 0.1 increments
inputs = arange(r_min, r_max+0.1, 0.1)
# compute targets
results = objective(inputs)
# create a line plot of input vs result
plt.plot(inputs, results)
plt.plot(value, eval_value, '.-', color='red')
# show the plot
plt.show()
```
